```python
import numpy as np
import matplotlib.pyplot as plt
from numpy import sin,cos,pi

Syt = 200 # Yield stress
S1 = 150 # Stress in direction 1 
S2 = 80 # Stress in direction 2 

Sv = np.sqrt(S1**2-(S1*S2)+S2**2)
a = 2**0.5*Syt # semi major axis of ellipse 
b = (2/3)**0.5*Syt # semi minor axis of ellipse 

theta = np.linspace(0,360*pi/180,720)

x = a*cos(theta) 
y = b*sin(theta) 

Sp1 = [200,160,80,60]
Sp2 = [-80,210,50,40]

rot_angle = pi/4 

xd = x*cos(rot_angle) - y*sin(rot_angle)
yd = x*sin(rot_angle) + y*cos(rot_angle)

xp = [Syt,Syt,0,-Syt,-Syt,0,Syt]
yp = [0,Syt,Syt,0,-Syt,-Syt,0]


px1 = [-300,300]
px2 = [0,0]

py1 = [0,0]
py2 = [-300,300]

plt.figure(figsize=(7.5,5))
plt.plot(xd,yd,label='Von Misses Region')
plt.plot(xp,yp, linestyle='--',label='Tresca Region')
plt.plot(px1,px2,linestyle='--',color='black')
plt.plot(py1,py2,linestyle='--',color='black')
plt.scatter(Sp1,Sp2)
plt.legend()
plt.show()
```
